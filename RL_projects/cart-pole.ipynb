# Imports

```python
import gymnasium as gym
import time
import numpy
```

# Create environment for game to run

```python
env = gym.make("CartPole-v1", render_mode = "human")
```

run the sim for a few episodes using random action

```python
episodes = 5

for episode in range(episodes):
    # resets the simulation at the start of episode
    observation, info = env.reset()

    done = False
    total_reward = 0.0

    while not done:
        # renders the current frame to the screen
        env.render()

        # take an action from random actions
        action = env.action_space.sample()

        # step the environment forward
        observation, reward, terminated, truncated, info = env.step(action)
        print(reward)

        total_reward += float(reward)

        # episode ends if pole falls or cart is of screen
        done = terminated or truncated

        # rest between episodes
        time.sleep(1)
    print(f"episode #{episode} done. Total reward: {total_reward}")
env.close()
```
